#### Advanced Financial Data Gathering

```python
import requests
from io import StringIO
```

```python
# Major Indices 
url = "https://finance.yahoo.com/markets/world-indices"
headers = {
    "User-Agent": "Edg/150.0.0.0"
}
html = requests.get(url, headers=headers).text

major_indices = pd.read_html(StringIO(html))[0]
```

```python
major_indices.head()
```

```python
ticker_list = major_indices['Symbol'].str.replace('^', '').str.lower().to_list()
ticker_list
```

```python
ticker_list = ['MSFT', 'TSLA', 'NVDA', 'GOOG', 'AMZN', 'AAPL', 'META', 'NFLX', 'GS', 'DIS']
```

```python
ticker_data = yf.Tickers(ticker_list)
ticker_data
```

```python
df = yf.download(ticker_list, start='2020-01-13', end='2021-03-10')
```

```python
len(ticker_list)
```

```python
df
```

```python
df.columns
```

```python
df.head()
```

```python
close = df.dropna(thresh=10, axis=1)['Close']
```

```python
close
```

```python
close.describe().T
```

```python
close.plot(figsize=(12,12), subplots=True);
```

```python
amzn = close['AMZN']
```

```python
amzn.resample('4ME').mean()
```

```python
amzn.shift(1)
```

```python
amzn_per_change = amzn / (amzn.shift(1) - 1)
amzn_per_change.plot(figsize=(12,6));
```

```python
amzn_log_returns_shift = np.log(amzn / amzn.shift(1))
```

```python
amzn_log_returns_shift.plot(figsize=(12,6));
```

```python
amzn.hist(bins=50);
```

```python
amzn_log_returns_shift.hist(bins=50, figsize=(10,10))
```
