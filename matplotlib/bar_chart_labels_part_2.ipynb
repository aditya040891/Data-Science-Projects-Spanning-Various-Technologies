## Bar Chart Labels with Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(42)
fig, ax = plt.subplots()
# Fake some data
companies = ("META", "AMZN", "NFLX", "GOOG")
y_pos = np.arange(len(companies))
performance = 4 + 10 * np.random.rand(len(companies))
err = np.random.rand(len(companies))

hbars = ax.barh(y_pos, performance, xerr=err, align='center')
ax.set_yticks(y_pos)
ax.set_yticklabels(companies)
ax.invert_yaxis() # labels top to bottom
ax.set_xlabel("Performance")
ax.set_title("FAANG Stock Performance Q1")

ax.bar_label(hbars, labels=['±%.2f' % e for e in err], padding=10, color='red', fontsize=14)
ax.set_xlim(right=18)
plt.show()
```
