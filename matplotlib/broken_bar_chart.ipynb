## Broken Bar Chart with Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
team_1 = [(110, 30), (150,10)]
team_2 = [(10, 50), (100, 20), (130, 10)]
```

```python
fig, ax = plt.subplots()
ax.broken_barh(team_1, (10,9), facecolor="tab:blue")
ax.broken_barh(team_2, (20,9), facecolors=("tab:red", "tab:green", "tab:orange"))
ax.set_ylim(5,35)
ax.set_xlim(0,200)
ax.set_xlabel("Minutes from Start")
ax.set_yticks([15, 25])
ax.set_yticklabels(["Team 1", "Team 2"])
ax.grid(True)
ax.annotate("Penalty", (61, 25), textcoords="axes fraction", xytext=(0.8, 0.9),
            arrowprops=dict(facecolor="black", shrink=0.05),
           fontsize=14, horizontalalignment="right", verticalalignment='top')
plt.show()
```
