## Plotting Categorical Variables with Matplotlib - Part 2

```python
import matplotlib.pyplot as plt
```

```python
boy = ["bored", "happy", "bored", "bored", "happy", "bored"]
girl = ["happy", "happy", "bored", "happy", "bored", "happy"]
```

```python
activity = ["breakfast", "reading", "games", "snack", "nap", "bathing"]
```

```python
fig, ax = plt.subplots()
ax.plot(activity, boy, label='boy')
ax.plot(activity, girl, label='girl')
ax.legend()
plt.show()
```
