#### If, Else, Elif statements

```python
x = 1
if x == 10:
    print('Hello')
```

```python
x == 10
```

```python
if 1 < 2:
    print("Line 1")
else:
    print("Line 2")
```

```python
password = '1234'
if password == '123':
    print("Welcome!")
else:
    print("Password Incorrect!")
```

```python
x = 3

if x == 2:
    print("first")
elif x == 3:
    print("middle")
else:
    print("last")
```
