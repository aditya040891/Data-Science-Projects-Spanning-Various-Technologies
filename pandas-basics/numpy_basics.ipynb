## NumPy Basics

* NumPy (Numpy) is a Linear Algebra Library for Python, the reason it is so important for Data Science with Python is that almost all of the libraries in the PyData Ecosystem rely on NumPy as one of their main building blocks.
* Numpy is also incredibly fast, as it has bindings to C libraries.
* We will only learn the basics of NumPy.

### Using NumPy

Importing it as a library:

```python
import numpy as np
```

Numpy has many built-in functions and capabilities. We will not cover them all but instead we will focus on some of the most important aspects of NumPy: vectors, arrays, matrices and number generation. Let's start by discussing arrays.

### NumPy Arrays

NumPy arrays are the main way we will use Numpy throughout the course. Numpy arrays essentially come in two flavours: vectors and matrices. Vectors are strictly 1-d arrays and matrices are 2-d (but you should note a matrix can still have only one row or one column).

Let's begin our introduction by exploring how to create NumPy arrays.

#### Random

Numpy also has lots of ways to create random number arrays:

##### rand

Create an array of the given shape and populate it with random samples from a uniform distribution over `[0,1]`.

```python
np.random.rand(5)
```

```python
np.random.rand(5,5)
```

##### randn

Return a sample (or samples) from the "standard normal" distribution. Unlike rand which is uniform:

```python
np.random.randn(2)
```

```python
np.random.randn(5,5)
```

##### randint

Return random integers from `low` (inclusive) to `high` (exclusive).

```python
np.random.randint(1,100)
```

```python
np.random.randint(1, 100, 10)
```

### Numpy Indexing and Selection

In this lecture we will discuss how to select elements or groups of elements from an array.

```python
# Creating sample array
arr = np.arange(0,11)
```

```python
# Show
arr
```

### Bracket Indexing and Selection

The simplest way to pick one or some elements of an array looks very similar to python lists:

```python
# Get a value at an index
arr[8]
```

```python
# Get values in a range
arr[1:5]  # Slice
```

```python
# Get values in a range
arr[0:5]
```

#### Broadcasting

Numpy arrays differ from a normal Python list because of their ability to broadcast:

```python
# Setting a value with index range (Broadcasting)
arr[0:5] = 100

# Show 
arr
```

```python
# Reset array
arr = np.arange(0,11)

# Show
arr
```

#### Broadcasting

```python
# Important notes on Slices
slice_of_arr = arr[0:6]

# Show slice
slice_of_arr
```

```python
# Change slice
slice_of_arr[:] = 99

# Show slice again
slice_of_arr
```

#### Broadcasting (Dangers)

Now note the changes also occurs in our original array!

```python
arr
```

Data is not copied, it's a view of the original array! This avoids memory problems!

#### Copying

```python
# To get a copy, need to be explicit
arr_copy = arr.copy()

arr_copy
```

```python
slice_of_arr_2 = arr_copy[0:5]
slice_of_arr_2
```

```python
slice_of_arr_2[:] = 1000
```

```python
slice_of_arr_2, arr_copy, arr
```

#### Indexing a 2D array (matrices)

The general format is **arr_2d[row][col]** or **arr_2d[row, col]**. I recommend usually the comma notation for clarity.

```python
arr_2d = np.array(([5,10,15], [20,25,30], [35,40,45]))

# Show
arr_2d
```

```python
# Indexing row
arr_2d[1]
```

```python
# Format is arr_2d[row][col] or arr_2d[row,col]

# Getting individual element value
arr_2d[1][2]
```

```python
# Getting individual element value
arr_2d[1,2]
```

```python
# 2D array slicing

# Shape (2,2) from top right corner
arr_2d[:2,1:]
```

```python
# Shape bottom row
arr_2d[2]
```

```python
# Shape bottom row
arr_2d[2, :]
```

#### Fancy Indexing

Fancy indexing allows you to select entire rows or columns out of order, to show this , let's quickly build out a numpy array:

```python
# Set up matrix
arr2d = np.zeros((10,10))
```

```python
# Length of array
arr_length = arr2d.shape[1]
```

```python
# Set up array
for i in range(arr_length):
    arr2d[i] = i

arr2d
```

```python
arr2d[[2,4,6,8]]
```

```python
# Allows in any order
arr2d[[6,4,2,7]]
```
