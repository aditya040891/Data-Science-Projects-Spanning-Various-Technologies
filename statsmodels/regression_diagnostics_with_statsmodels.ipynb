## Regression Diagnostics with Statsmodels

```python
import statsmodels.api as sm
from statsmodels.compat import lzip
import statsmodels.formula.api as smf
import statsmodels.stats.api as sms

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = sm.datasets.get_rdataset('Guerry', 'HistData').data
```

```python
df
```

```python
df.columns
```

```python
res = smf.ols("Lottery~Literacy + np.log(Pop1831)", data=df).fit()
```

```python
res.summary()
```

### Normality of Residuals
#### Jarque-Bera Test:

```python
name = ["Jarque-Bera", "Chi^2 two-tail prob.", "Skew", "Kurtosis"]
```

```python
test = sms.jarque_bera(res.resid)
lzip(name, test)
```

#### Influence Tests

```python
from statsmodels.stats.outliers_influence import OLSInfluence
```

```python
test_class = OLSInfluence(res)
```

```python
test_class.dfbeta[5:, :]
```

```python
from statsmodels.graphics.regressionplots import plot_leverage_resid2
```

```python
fig, ax = plt.subplots(figsize=(10,6))
fig = plot_leverage_resid2(res, ax=ax)
```

### Multicollinearity

```python
np.linalg.cond(res.model.exog)
```

### Heteroskedasticity

```python
name = ['Lagrange multiplier stat', 'p-value', 'f-value', 'f g-value']
test = sms.het_breuschpagan(res.resid, res.model.exog)
lzip(name, test)
```

```python
name = ['F stat', 'p-value']
test = sms.het_goldfeldquandt(res.resid, res.model.exog)
lzip(name, test)
```

#### Linearity

```python
name = ['t value', 'p value']
test = sms.linear_harvey_collier(res)
lzip(name, test)
```
